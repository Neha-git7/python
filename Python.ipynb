1. What is Python, and why is it popular?

Python is a high-level, interpreted, general-purpose programming language. It is popular because it has simple and readable syntax, is easy to learn, supports many libraries and frameworks, and can be used for web development, data analysis, artificial intelligence, machine learning, automation, and many other applications.

2. What is an interpreter in Python?

An interpreter is a program that executes Python code. It reads and executes Python instructions and reports errors during execution. This allows Python programs to be tested and run easily without first compiling the entire program into machine code.

3. What are pre-defined keywords in Python?

Keywords are reserved words that have a special meaning in Python. They are part of the Python language syntax and cannot normally be used for other purposes. Examples include if, else, for, while, def, class, True, False, and None.

4. Can keywords be used as variable names?

No. Python keywords cannot be used as variable names because they have special meanings in the language.

5. What is mutability in Python?

Mutability refers to whether an object's value can be changed after the object is created. A mutable object can be changed, while an immutable object cannot be changed after creation. Lists and dictionaries are mutable, whereas strings, integers, and tuples are immutable.

6. Why are lists mutable, but tuples are immutable?

Lists are designed to allow their elements to be modified, added, or removed after creation. Tuples are designed to remain unchanged after creation. Therefore, we can modify a list but cannot directly change an element of a tuple.

7. What is the difference between == and is operators?

The == operator checks whether two objects have equal values, while the is operator checks whether two variables refer to the same object in memory.

8. What are logical operators in Python?

Logical operators are used to combine or reverse conditions. Python has three logical operators:

and – returns True if both conditions are True.
or – returns True if at least one condition is True.
not – reverses the result of a condition.
9. What is type casting in Python?

Type casting is the process of converting a value from one data type to another. Common functions include int(), float(), str(), and bool().

10. What is the difference between implicit and explicit type casting?

Implicit type casting is automatically performed by Python when it safely converts one data type into another.

Explicit type casting is performed manually by the programmer using functions such as int(), float(), and str().

11. What is the purpose of conditional statements in Python?

Conditional statements are used to make decisions based on conditions. They allow a program to execute different blocks of code depending on whether a condition is true or false. Python uses if, elif, and else.

12. How does the elif statement work?

elif means "else if". It allows us to check another condition when the previous if or elif condition is false.

13. What is the difference between for and while loops?

A for loop is generally used when we want to iterate over a sequence or when the number of iterations is known.

A while loop is used when we want to repeat a block of code as long as a condition remains true, especially when the number of iterations is not known in advance.

14. When is a while loop more suitable than a for loop?

A while loop is more suitable when the number of repetitions is not known beforehand and depends on a condition.

For example, a program that repeatedly asks a user to enter a password until the correct password is entered can use a while loop.

14. Write a program to reverse a string using a while loop.

```python
text = input("Enter a string: ")

reversed_text = ""
i = len(text) - 1

while i >= 0:
    reversed_text += text[i]
    i -= 1

print("Reversed string:", reversed_text)
```

15. Write a Python program to calculate the factorial of a number provided by the user using a while loop.

```python
number = int(input("Enter a number: "))

factorial = 1
i = 1

while i <= number:
    factorial *= i
    i += 1

print("Factorial:", factorial)
```
